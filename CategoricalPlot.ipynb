## Categorical Plot

```python
import matplotlib as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df= pd.read_csv("tips.csv")
df
```

```python
sex_ = sns.countplot(x = df['sex'], hue = df['smoker'])
```

```python
smoker_ = sns.countplot(x = df["smoker"])
```

```python
sex_ = sns.countplot(x = df['sex'])
```

```python
sns.barplot(x = df["sex"], y = df["total_bill"])
```

```python
sns.barplot(x = df["sex"], y = df["total_bill"], estimator = np.sum) 
```

```python
sns.barplot(x = df["sex"], y = df["tip"], estimator = np.mean) 
```

```python
sns.boxplot(x = "tip", y = "day", data = df , palette ='rainbow')
```

```python
sns.violinplot( x = "day", y = "total_bill" , data = df)
```

```python
yupp = sns.stripplot(x = "day" , y = "total_bill", data = df)
```

```python
yoooo = sns.swarmplot(data = df, x = "day" , y = "total_bill")
```

```python
sns.violinplot(x = "day", y = "tip", data = df)
sns.swarmplot(x = "day", y = "tip", data = df)
```
