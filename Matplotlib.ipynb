```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
```

```python
x = np.linspace(0, 10, 100)
y = x ** 2
```

# Basic commands in matplotlib

```python
plt.plot(x, y)
plt.title("Square No.")
plt.xlabel('YOOOO')
plt.ylabel('YUUUUP')
```

# Subplots in Matplotlib

```python
plt.subplot(2,2,1)
plt.plot(y,y)
plt.subplot(2,2,2)
plt.plot(x**y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x,y)
```

```python
# Object oriented way
```

```python
fig = plt.figure(figsize=(8,8), dpi = 100)
ax1 = fig.add_axes([0.1, 0.1 , 1 , 1])
ax1.plot(x,y)
ax2 = fig.add_axes([0.5,0.31,0.3,0.3])
ax2.plot(x,y)
```

## Saving the Plot

```python
fig = plt.figure(figsize = (10,10), dpi = 100)
ax3 = fig.add_axes([0.1,0.1,0.8,0.8])
plt.plot(x,y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title('HEllo MOTO')
plt.savefig("basic.png")
```
