```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
print("========== Sales Data ==========")
print("\n Sales data Shape:", sales_data.shape) 
print("\n For First three restaurants:\n", sales_data[:3])
```

## Total Seals PEr Year

```python
print(np.sum(sales_data, axis=0))
yearly_total = np.sum(sales_data[:,1:], axis=0)
print("\n Yearly Total Sales: \n", yearly_total)
total_sum = np.sum(yearly_total)
print("\n Total Sales: \n", total_sum)

# minimum sales per restaurant
min_sales = np.min(sales_data[:,1:], axis=1)
print("\n Minimum Sales per Restaurant: \n", min_sales)

# maximum sales per restaurant
max_sales = np.max(sales_data[:,1:], axis=1)
print("\n Maximum Sales per Restaurant: \n", max_sales)


## Average Sales per Restaruants
avg_sales = np.mean(sales_data[:,1:], axis=1)
print('\n Avg. Sales Per Restaurant: \n', avg_sales)
```
